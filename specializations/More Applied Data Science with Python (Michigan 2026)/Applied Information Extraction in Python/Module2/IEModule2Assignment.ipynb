````python
import json
import numpy as np
from hmmlearn import hmm
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

with open("list_of_real_usa_addresses.json") as f:
    addresses_json = json.load(f)

def extract_features(token):
    return {
        "word": token,
        "length": len(token),
        "is_digit": token.isdigit(),
        "is_upper": token.isupper()
    }

def replace_token_with_features():
    old_X, _ = preprocess()
    return [[extract_features(t) for t in seq] for seq in old_X]

def predict(X, y):
    idx = np.arange(len(X))
    train, test = train_test_split(idx, test_size=0.2, random_state=42)

    def key(f):
        return tuple(sorted(f.items()))

    # Build vocabulary from training data only
    vocab = {
        k: i for i, k in enumerate({
            key(f) for i in train for f in X[i]
        })
    }

    unk = len(vocab)
    V = len(vocab) + 1  # Reserve one column for unknown tokens

    def encode(seq):
        return np.array(
            [vocab.get(key(f), unk) for f in seq],
            dtype=int
        ).reshape(-1, 1)

    le = LabelEncoder().fit([tag for i in train for tag in y[i]])
    K = len(le.classes_)
    alpha = 0.1

    start = np.full(K, alpha)
    trans = np.full((K, K), alpha)
    emit = np.full((K, V), alpha)

    for i in train:
        obs = encode(X[i]).ravel()
        states = le.transform(y[i])

        if len(states) == 0:
            continue

        start[states[0]] += 1

        for a, b in zip(states[:-1], states[1:]):
            trans[a, b] += 1

        for o, s in zip(obs, states):
            emit[s, o] += 1

    model = hmm.CategoricalHMM(n_components=K, n_iter=100, init_params="")
    model.n_features = V
    model.startprob_ = start / start.sum()
    model.transmat_ = trans / trans.sum(axis=1, keepdims=True)
    model.emissionprob_ = emit / emit.sum(axis=1, keepdims=True)

    y_pred = [le.inverse_transform(model.predict(encode(X[i]))) for i in test]

    return y_pred, [y[i] for i in test]

def compute_accuracy():
    X = replace_token_with_features()
    _, y = preprocess()

    y_pred, y_test = predict(X, y)

    correct = sum(
        p == t
        for pred, true in zip(y_pred, y_test)
        for p, t in zip(pred, true)
    )
    total = sum(len(seq) for seq in y_test)

    return correct / total if total else 0.0

print("HMM accuracy:", compute_accuracy())
````

The increase to 95% accuracy is primarily explained by correcting the observation encoding and ensuring that the HMM estimates valid emission probabilities. However, an important distinction is that the corrected code fixes a representation/indexing problem; it does not make the HMM equivalent to a CRF. Let's derive exactly what changed mathematically.

## 1. What the HMM is learning

Suppose an address consists of tokens

\\[ X=(x_1,x_2,\ldots,x_T) \\]

and their named-entity labels are

\\[ Y=(y_1,y_2,\ldots,y_T), \\]

where labels might include `B-LOC`, `I-LOC`, `B-ORG`, and `O`.

A first-order Hidden Markov Model assumes

\\[ P(Y,X)=P(y_1) \prod\_{t=2}^{T}P(y_t\mid y\_{t-1}) \prod\_{t=1}^{T}P(x_t\mid y_t). \\]

There are two main components:

- Transition probabilities: \\(P(y_t\mid y\_{t-1})\\), which describe how entity labels follow one another.
- Emission probabilities: \\(P(x_t\mid y_t)\\), which describe how observations are associated with entity labels.

The HMM finds the most likely label sequence using the Viterbi algorithm:

\\[ \hat{Y}=\arg\max_Y P(Y\mid X). \\]

Since \\(P(X)\\) is constant for a given input, this is equivalent to maximizing \\(P(Y,X)\\).

## 2. The major change: observation encoding

Your earlier HMM code contained this line:

```
X_enc = [vec.transform(seq).toarray().argmax(axis=1) for seq in X]
```

This was a serious problem.

### 2.1 What `DictVectorizer` actually produces

Consider these feature dictionaries:

```
{"word": "Boston", "length": 6, "is_digit": False, "is_upper": False}{"word": "12345", "length": 5, "is_digit": True, "is_upper": False}
```

`DictVectorizer` converts categorical and numerical features into a vector of feature values. Conceptually, a sparse representation might look like this:

Illustrative feature vectors

Boston
0 0 1 0 6 0

12345
0 1 0 0 5 0

Illustration only; the actual vector layout depends on the fitted vocabulary and feature names.

The crucial point is that a feature vector contains multiple dimensions. Calling

```
.toarray().argmax(axis=1)
```

reduces each vector to the index of its largest element.

Mathematically, if the feature vector for token \\(x\\) is

\\[ \phi(x)=(\phi_1(x),\ldots,\phi_d(x)), \\]

the code retains only

\\[ c(x)=\arg\max_j\phi_j(x). \\]

It discards nearly all the information in \\(\phi(x)\\). Many different tokens can therefore map to the same integer. That integer is not a faithful representation of the token's complete features.

### 2.2 What the corrected code does instead

The corrected implementation defines:

```
def key(f):    return tuple(sorted(f.items()))
```

For example, a token's features become a tuple resembling

```
(    ("is_digit", False),    ("is_upper", False),    ("length", 6),    ("word", "Boston"))
```

Each distinct complete feature combination receives its own integer ID.

Mathematically, define a mapping

\\[ g:\mathcal{F}\rightarrow\\{0,1,\ldots,V-2\\}, \\]

where \\(\mathcal{F}\\) is the set of distinct feature dictionaries found in the training data. Then

\\[ o_t=g(\phi(x_t)). \\]

This mapping is one-to-one over the observed feature combinations in the training vocabulary. Unlike `argmax`, it does not collapse different combinations merely because one feature has the same maximum index.

Consequently, the HMM learns emissions from more meaningful observation categories.

Important qualification: This is not a general solution for rich feature modeling. It treats the entire dictionary as one categorical observation. A token not seen in training maps to the unknown-token ID.

## 3. The unknown-token column fixes the indexing error

Your earlier error was:

```
IndexError: index 1114 is out of bounds for axis 1 with size 747
```

The underlying issue was that the code generated an observation ID that did not have a corresponding column in the emission matrix.

The corrected code uses:

```
unk = len(vocab)V = len(vocab) + 1
```

Suppose the training vocabulary contains \\(M=747\\) distinct observations.

Their valid IDs are

\\[ 0,1,\ldots,746. \\]

The unknown observation receives ID

\\[ o\_{\mathrm{UNK}}=747. \\]

The emission matrix must therefore contain \\(748\\) columns, with indices \\(0,\ldots,747\\).

In general, the dimensions are

\\[ B\in\mathbb{R}^{K\times V}, \qquad V=M+1, \\]

where \\(K\\) is the number of label classes.

This ensures every encoded observation has a valid emission-matrix index.

The unknown column is initialized by smoothing and receives observations only if unknown tokens occur in training sequences. Since the vocabulary is built from the training data, unknown observations will typically occur only in the test set; their emission distributions will therefore remain determined by smoothing.

## 4. Additive smoothing prevents zero probabilities

The corrected code initializes the counts with

```
alpha = 0.1start = np.full(K, alpha)trans = np.full((K, K), alpha)emit = np.full((K, V), alpha)
```

This is additive (Laplace-style) smoothing, also known as a symmetric Dirichlet pseudocount approach.

Without smoothing, maximum-likelihood estimates of the emission probabilities would be

\\[ \hat P(o\mid y)= \frac{C(y,o)}{\sum\_{o'}C(y,o')}, \\]

where \\(C(y,o)\\) is the number of times observation \\(o\\) occurs with label \\(y\\).

If \\(C(y,o)=0\\), the estimated probability is zero. A single zero probability can make an entire candidate path have probability zero.

With smoothing,

\\[ \boxed{ \hat P(o\mid y)= \frac{C(y,o)+\alpha} {\sum\_{o'}C(y,o')+\alpha V} } \\]

and similarly for transitions,

\\[ \boxed{ \hat P(y'\mid y)= \frac{C(y,y')+\alpha} {\sum_z C(y,z)+\alpha K} }. \\]

The initial-state probabilities are estimated as

\\[ \boxed{ \hat P(y)= \frac{C\_{\mathrm{start}}(y)+\alpha} {N\_{\mathrm{seq}}+\alpha K} }. \\]

Here \\(N\_{\mathrm{seq}}\\) is the number of training sequences.

Smoothing prevents zero probabilities and makes the model more robust to rare transitions and emissions. It does not guarantee that every unknown observation is correctly classified.

## 5. The model now estimates probabilities from training labels

For each training sequence, the code increments counts:

```
start[states[0]] += 1for a, b in zip(states[:-1], states[1:]):    trans[a, b] += 1for o, s in zip(obs, states):    emit[s, o] += 1
```

These operations estimate three distributions:

| Quantity | What is counted                        |
| -------- | -------------------------------------- |
| `start`  | The first label in each sequence       |
| `trans`  | Consecutive pairs of labels            |
| `emit`   | Observation IDs associated with labels |

The counts are then normalized:

```
model.startprob_ = start / start.sum()model.transmat_ = trans / trans.sum(axis=1, keepdims=True)model.emissionprob_ = emit / emit.sum(axis=1, keepdims=True)
```

This gives each transition-matrix row and emission-matrix row a sum of one, as required for probability distributions.

Notice that the code estimates these probabilities directly from the known training labels. It is a supervised HMM, even though `hmmlearn` also supports unsupervised estimation.

## 6. Why 95% accuracy is plausible

Address data often contains strong regularities. For example:

- Numeric tokens frequently correspond to house numbers or ZIP codes.
- Capitalized words may occur in place names.
- Certain tokens tend to follow specific entity labels.
- Entity labels often follow consistent BIO transition patterns.

The HMM can exploit the learned transition and emission distributions to predict these patterns.

For example, if the model has learned that `B-LOC` is frequently followed by `I-LOC`, the transition probability

\\[ P(I\text{-LOC}\mid B\text{-LOC}) \\]

may be relatively high. If a particular observation also has a high emission probability under `B-LOC`, the combined path can be favored by Viterbi decoding.

However, 95% token accuracy does not necessarily mean 95% of address entities are correctly identified. If `O` is the dominant label, the model may achieve high token accuracy while missing many complete entities.

# How Conditional Random Fields (CRFs) work for Named Entity Recognition

Your code uses a linear-chain Conditional Random Field (CRF) to identify named entities in address sequences. Unlike the HMM you implemented earlier, a CRF directly models the conditional probability of an entire label sequence given the observed tokens and their features.

We will derive the mathematical model, explain how your four features contribute to it, show how the parameters are learned using L-BFGS, and connect each mathematical component to your Python code.

## 1. The NER problem: from tokens to labels

Suppose your input address is:

```
123 Main Street Boston
```

The preprocessing function `preprocess()` is assumed to return token sequences and corresponding BIO labels.

For example:

Input sequence

123 Main Street Boston

Possible BIO labels

123    Main   Street  Boston 
B-STR  B-STR  I-STR   B-LOC

Illustrative labels only; the actual labels depend on your dataset's annotation scheme.

Let the input sequence be

\\[ X=(x_1,x_2,\ldots,x_T) \\]

and the associated label sequence be

\\[ Y=(y_1,y_2,\ldots,y_T). \\]

Here:

- \\(T\\) is the number of tokens.
- \\(x_t\\) is the token at position \\(t\\).
- \\(y_t\\) is its entity label, such as `B-LOC`, `I-LOC`, or `O`.

The CRF learns to predict

$$ \boxed{\hat Y=\arg\max\limits_Y P(Y\mid X)}$$. 

The important point is that it predicts the whole sequence jointly, rather than classifying every token independently.

## 2. How your feature extraction works mathematically

Your function is:

```
def extract_features(token):    return {        "word": token,        "length": len(token),        "is_digit": token.isdigit(),        "is_upper": token.isupper()    }
```

For every token \\(x_t\\), it produces a feature representation

\\[ \phi(x_t)= \begin{bmatrix} \text{word}(x_t)\\\ \text{length}(x_t)\\\ \text{isDigit}(x_t)\\\ \text{isUpper}(x_t) \end{bmatrix}. \\]

These are heterogeneous features: the word is categorical, the length is numerical, and the final two features are Boolean.

For example, the features for `"123"` are:

```
{    "word": "123",    "length": 3,    "is_digit": True,    "is_upper": False}
```

The features for `"BOSTON"` are:

```
{    "word": "BOSTON",    "length": 6,    "is_digit": False,    "is_upper": True}
```

The CRF library handles these dictionaries and learns weights associated with feature values. You do not need to manually convert the dictionaries into numerical vectors.

### What does each feature tell the CRF?

| Feature    | Mathematical role   | Potential usefulness                                               |
| ---------- | ------------------- | ------------------------------------------------------------------ |
| `word`     | Token identity      | Learns associations such as `"Boston"` with `B-LOC`                |
| `length`   | Token length        | May help distinguish numbers, abbreviations, or common token types |
| `is_digit` | Numeric indicator   | Helps recognize house numbers and ZIP codes                        |
| `is_upper` | Uppercase indicator | Helps identify capitalized place names and abbreviations           |

These features provide evidence about the label. None of them independently determines the correct label.

For example, `"Boston"` and `"BOSTON"` have different `word` values, even though both might represent the same location. Likewise, `is_upper` alone cannot distinguish a location from an organization.

## 3. The central idea: CRF assigns a score to a complete label sequence

A linear-chain CRF defines a real-valued score for a candidate sequence \\(Y\\) given the input \\(X\\).

The score has two components:

1. State scores: How well does each label match the features of its token?
2. Transition scores: How compatible are consecutive labels?

The total score is

$$ \boxed{ S_\theta(X,Y)= \sum\limits_{t=1}^{T}\sum\limits_k w_k f_k(y_{t-1},y_t,X,t) } $$

where:

- \\(f_k\\) is a feature function.
- \\(w_k\\) is its learned weight.
- \\(\theta\\) denotes all model weights.
- \\(y_{t-1}\\) and \\(y_t\\) are consecutive labels.

The feature functions may depend on the current label, the previous label, and the observed tokens.

A common equivalent decomposition is

$$ S\_\theta(X,Y)= \sum\limits_{t=1}^{T} \left[ \underbrace{s\_\theta(y_t,X,t)}_{\text{state score}} + \underbrace{a_\theta(y_{t-1},y_t)}_{\text{transition score}} \right]$$.

Here, \\(s\_\theta\\) measures the evidence for assigning a label to a token, while \\(a\_\theta\\) measures the compatibility of two adjacent labels.

### 3.1 State scores: which label fits a token?

Suppose the CRF has three possible labels:

\\[ \mathcal{Y}=\\{\text{B-LOC},\text{I-LOC},\text{O}\\}. \\]

For a token `"Boston"`, the model may learn weights that make `B-LOC` a strong candidate because the token identity `"Boston"` has been associated with that label in the training data.

A simplified state score might be

\\[ \begin{aligned} s(\text{B-LOC},x_t) ={}&w_1\mathbf{1}[x_t=\text{"Boston"}]\\\ &+w_2\mathbf{1}[\operatorname{isUpper}(x_t)]\\\ &+w_3\operatorname{length}(x_t). \end{aligned} \\]

The indicator function is defined as

\\[ \mathbf{1}[A]= \begin{cases} 1,& A\text{ is true},\\\ 0,& A\text{ is false}. \end{cases} \\]

This is an illustrative feature-score equation, not a claim that these are the exact features or weights learned by your installed library.

For `"Boston"`, the word-identity feature activates. For `"BOSTON"`, the uppercase feature activates as well. The model learns from data how strongly these features should influence each label.

In general, each label has its own learned feature weights:

\\[ s_\theta(y,x_t)=\mathbf{w}_y^\top\phi(x_t). \\]

Thus, the same token features can provide different evidence for different labels.

### 3.2 Transition scores: which labels should follow one another?

NER labels have sequential structure. For example, an `I-LOC` label typically follows a `B-LOC` or `I-LOC` label, whereas `O` denotes a token outside the entity.

A transition-score matrix can be written as

\\[ A= \begin{bmatrix} a_{BB}&a_{BI}&a_{BO}\\\ a_{IB}&a_{II}&a_{IO}\\\ a_{OB}&a_{OI}&a_{OO} \end{bmatrix}, \\]

where the rows represent the previous label and the columns represent the current label. The actual matrix would have one row and column for every label in your dataset, not just three.

A large transition score encourages a particular label pair; a small or negative score discourages it.

For example,

\\[ a(\text{B-LOC},\text{I-LOC}) > a(\text{B-LOC},\text{O}) \\]

would favor continuing a location entity over ending it immediately, all else being equal.

However, a standard CRF does not automatically forbid invalid BIO transitions. Such constraints must be imposed explicitly if required. Your `all_possible_transitions=True` option makes the library include transitions between all label pairs in the model's feature space; it does not enforce BIO legality.

## 4. From scores to probabilities: the partition function

A score alone is not a probability. The CRF converts scores into a normalized conditional probability distribution using the exponential function.

For any candidate label sequence \\(Y\\),

\\[ \boxed{ P_\theta(Y\mid X)= \frac{\exp(S_\theta(X,Y))} {Z_\theta(X)} } \\]

where the partition function is

\\[ \boxed{ Z_\theta(X)= \sum\limits_{Y'\in\mathcal{Y}^{T}} \exp(S_\theta(X,Y')). } \\]

The summation includes every possible label sequence of length \\(T\\).

Why is this normalization necessary? It ensures that

\\[ \sum\limits_{Y\in\mathcal{Y}^{T}}P_\theta(Y\mid X)=1. \\]

The partition function depends on the input \\(X\\), but it is shared by all candidate label sequences for that same input.

### A numerical example

Suppose, for a short sequence, three candidate label sequences have scores

\\[ S_1=2,\qquad S_2=1,\qquad S_3=0. \\]

Their unnormalized probabilities are

\\[ e^2\approx7.389,\quad e^1\approx2.718,\quad e^0=1. \\]

The partition function, assuming these are the only candidates in this illustrative example, is

\\[ Z=e^2+e^1+e^0\approx11.107. \\]

Therefore,

\\[ P_1\approx0.665,\qquad P_2\approx0.245,\qquad P_3\approx0.090. \\]

The highest-scoring sequence has the highest probability. In a real CRF, all possible sequences contribute to \\(Z\\), and dynamic programming avoids explicitly enumerating them.

### Why this differs from an HMM

An HMM models the joint distribution of observations and hidden labels, commonly as

\\[ P(X,Y)=P(y_1) \prod_{t=2}^{T}P(y_t\mid y_{t-1}) \prod_{t=1}^{T}P(x_t\mid y_t). \\]

A CRF directly models

\\[ P(Y\mid X). \\]

Consequently, the CRF does not need to define a generative probability distribution \\(P(X)\\). It can use arbitrary observed-token features, including word identity, token length, capitalization, and other overlapping or correlated features.

This is a major reason CRFs are often effective for NER.

## 5. How the CRF learns its weights from labeled examples

Your training call is:

```
crf.fit(X_train, y_train)
```

At this stage, the CRF already knows the feature values and the correct labels for the training sequences. It must learn the weights \\(\theta\\) that assign high probability to the correct label sequences.

### 5.1 Conditional log-likelihood

Suppose the training dataset contains \\(N\\) labeled sequences:

\\[ \mathcal D=\\{(X^{(i)},Y^{(i)})\\}_{i=1}^{N}. \\]

The conditional log-likelihood of the training data is

\\[ \mathcal L(\theta)= \sum\limits_{i=1}^{N} \log P_\theta(Y^{(i)}\mid X^{(i)}). \\]

Substituting the CRF probability formula,

\\[ \begin{aligned} \mathcal L(\theta) &=\sum\limits_{i=1}^{N} \log\frac{\exp(S_\theta(X^{(i)},Y^{(i)}))} {Z_\theta(X^{(i)})}\\\ &=\sum\limits_{i=1}^{N} \left[ S_\theta(X^{(i)},Y^{(i)}) -\log Z_\theta(X^{(i)}) \right]. \end{aligned} \\]

This equation explains the training objective.

- The first term rewards weights that give the correct label sequence a high score.
- The second term accounts for all possible label sequences, preventing the model from arbitrarily increasing scores without improving the probability of the correct sequence.

Training seeks parameters that make the observed correct sequences probable relative to competing sequences.

### 5.2 What happens to a feature weight?

Consider a feature indicating that the token is `"Boston"` and the current label is `B-LOC`.

Let \\(C_k(X,Y)\\) be the number of times feature \\(k\\) is activated in a sequence. The score can be written as

\\[ S_\theta(X,Y)=\sum\limits_k w_k C_k(X,Y). \\]

Ignoring regularization for a moment, the derivative of the log-likelihood with respect to a feature weight is

\\[ \boxed{ \frac{\partial\mathcal L}{\partial w_k} = \sum\limits_{i=1}^{N} \left[ C_k(X^{(i)},Y^{(i)}) - \mathbb E_{P_\theta(Y\mid X^{(i)})} C_k(X^{(i)},Y) \right]. } \\]

This is one of the most important equations for understanding CRF training.

The gradient is the difference between:

1. The feature count in the correct, observed labels.
2. The expected feature count under the model's current distribution over possible labels.

If the model assigns insufficient probability to correct `Boston`–`B-LOC` associations, optimization can increase the relevant weight. If the model overpredicts that association relative to the training labels, optimization can decrease the weight.

Notice that the second term is an expectation over label sequences, not merely a comparison with the single best predicted sequence.

## 6. Regularization: the meaning of `c1=0.1` and `c2=0.1`

Your model is initialized with:

```
crf = sklearn_crfsuite.CRF(    algorithm="lbfgs",    c1=0.1,    c2=0.1,    max_iterations=100,    all_possible_transitions=True)
```

The `c1` and `c2` parameters control regularization. In `sklearn-crfsuite`, they correspond to L1 and L2 regularization strengths.

A simplified negative-log-likelihood objective is

\\[ J(\theta)= -\mathcal L(\theta) +c_1\\|\theta\\|_1 +\frac{c_2}{2}\\|\theta\\|_2^2. \\]

Here,

\\[ \\|\theta\\|_1=\sum\limits_k|w_k|, \qquad \\|\theta\\|_2^2=\sum\limits_kw_k^2. \\]

The precise objective scaling and treatment of bias parameters depend on the underlying implementation, but this equation captures the regularization effects.

### L1 regularization: `c1=0.1`

The L1 penalty discourages unnecessary weights and can drive some weights exactly to zero.

This can produce a more compact model, especially when many word-specific or transition features are weakly informative.

### L2 regularization: `c2=0.1`

The L2 penalty discourages excessively large weights. It generally shrinks weights toward zero without forcing as many of them to become exactly zero.

This helps control overfitting, which can be important when the dataset contains many rare words.

Both penalties introduce a trade-off: stronger regularization may improve generalization but can also reduce the model's ability to fit useful patterns.

## 7. Why the algorithm is called L-BFGS

The option

```
algorithm="lbfgs"
```

selects the Limited-memory Broyden–Fletcher–Goldfarb–Shanno optimization algorithm.

L-BFGS is a numerical optimization method that uses the objective value and its gradient to find good parameter values without storing a full Hessian matrix.

Conceptually, each optimization iteration:

1. Evaluates the regularized negative log-likelihood.
2. Computes gradients with respect to the model weights.
3. Uses gradient and curvature information from recent iterations to choose an update direction.
4. Updates the weights and repeats until convergence or an iteration limit is reached.

The update can be viewed abstractly as

\\[ \theta_{m+1}=\theta_m+\eta_m d_m, \\]

where \\(d_m\\) is the optimizer's search direction and \\(\eta_m\\) is a step length selected by its line-search procedure.

This is not simply ordinary gradient descent: L-BFGS uses an approximation to second-order curvature information to improve its search directions.

Your setting

```
max_iterations=100
```

limits the number of optimization iterations. It does not guarantee convergence within 100 iterations.

## 8. Prediction: how the CRF chooses labels

Your prediction code is:

```
y_pred = crf.predict(X_test)
```

The CRF chooses the most probable complete label sequence:

\\[ \boxed{ \hat Y=\arg\max\limits_{Y\in\mathcal Y^T} P_\theta(Y\mid X). } \\]

Because the partition function \\(Z_\theta(X)\\) is the same for every candidate \\(Y\\) for a fixed input, this is equivalent to

\\[ \hat Y=\arg\max\limits_Y S_\theta(X,Y). \\]

The Viterbi algorithm finds the maximum-scoring sequence efficiently by dynamic programming.

Let \\(D_t(y)\\) denote the maximum score of a partial sequence ending in label \\(y\\) at position \\(t\\). Its recurrence is

\\[ D_t(y)=s_\theta(y,X,t) +\max_{y'\in\mathcal Y} \left[D_{t-1}(y')+a_\theta(y',y)\right]. \\]

At every token position, the algorithm considers the possible preceding labels, retains the best score for each current label, and records the predecessor that achieved that score. Backtracking then recovers the best full sequence.

This is how the CRF combines evidence from individual tokens with information about neighboring labels.

## 10. How `compute_accuracy()` calculates accuracy

Your code returns:

```

```

If the test set contains \\(M\\) total tokens, token-level accuracy is

\\[ \boxed{ \operatorname{Accuracy} =\frac{\sum\limits_{t=1}^{M} \mathbf 1[\hat y_t=y_t]}{M}. } \\]

Suppose 950 out of 1,000 test tokens receive the correct labels. The accuracy is

\\[ \frac{950}{1000}=0.95=95\\%. \\]

However, this does not mean that 95% of the named entities are fully correct. An entity may span several tokens, and a single incorrect BIO tag can cause the entire entity span to be wrong.

For a serious NER evaluation, report entity-level precision, recall, and F1 in addition to token accuracy. Also, if addresses from the same source or template appear in both training and testing, a random sequence split may overestimate performance; a grouped or source-separated split can provide a more realistic assessment.

## 11. The essential difference between your HMM and CRF

| Property                         | HMM                                                     | Linear-chain CRF                                 |
| -------------------------------- | ------------------------------------------------------- | ------------------------------------------------ |
| Main model                       | \\(P(X,Y)\\)                                            | \\(P(Y\mid X)\\)                                 |
| Token evidence                   | Emission probabilities                                  | Arbitrary feature functions                      |
| Sequence structure               | Transition probabilities                                | Transition feature weights                       |
| Feature handling                 | Standard categorical HMM requires discrete observations | Can directly use rich feature dictionaries       |
| Training in your implementations | Counts and smoothing                                    | Conditional likelihood optimization              |
| Sequence prediction              | Viterbi                                                 | Viterbi                                          |
| Main limitation                  | Strong generative assumptions and observation modeling  | Can overfit with sparse features or limited data |

The key takeaway is that a CRF jointly learns how token features support labels and how labels relate to neighboring labels, then selects the most probable complete label sequence. That combination makes it particularly useful for BIO-style NER.
