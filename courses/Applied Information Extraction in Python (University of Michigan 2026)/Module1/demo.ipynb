# Demo: Using regex expressions and examining output

```python
import re
```

The following code shows how you can use regex patterns to extract data from a line.

```python
textline = """
     gibberish, 2024-06-30, 30 Jun 2024, gibberish, Jun 30, 2024, 06/30/2024, 30-6-2024
"""

reg = re.compile(r"""
    (\d{4})-(\d{2})-(\d{2})
    """, re.VERBOSE)

dates = re.findall(reg, textline)

print(dates)
```

To add more patterns a reg expression can recognize, simply add a `|` and input another regex expression like so. 

```python
reg = re.compile(r"""
    (\d{4})-(\d{2})-(\d{2}) |           # Matches 2024-06-30
    (\d{1,2})\s([A-Za-z]{3})\s(\d{4}) # Matches 30 Jun 2024
    """, re.VERBOSE)

dates = re.findall(reg, textline)

print(dates)
```

You may notice some empty strings in these tuples now. Do not worry about these, as these are easily filtered out.
