# Task 1: Create reg expression

These are possible formats a phone number could be stored in:

Parentheses: (409) 339-9584

Dashes: 943-641-7270

Dots: 574.874.7163

Spaces: 175 335 3625

International: +1-536-234-1765

Plain: 6565865784

For this task, you must define the 'reg' expression such that the given code is able to extract phone numbers in every one of these formats.

When we use reg.findall, we want to obtain a list of 3 strings containing the digits of each phone number separated by section. For example, if we have:

`str = John Smith: 123 456 7890`

We want `re.findall(reg, str)` on this line using our regex expression to obtain a list that contains the following:

['123','456','7890']

Look up how to use regex 'capturing groups' to accomplish this. The demo notebook associated with this assignment also contains hints on how you can accomplish this task. 

Assign your reg expression to `reg` in `task_1_solution`.

We use your reg expression in `extract_contents` to extract the contents of `lines` into a list where each entry in the list formatted like so:

`['First Last','000', '000', '0000']`

````python
import re


def extract_contents(reg):
    extracted_contents = []
    for line in lines:

        name, number = line.strip().split(':')
        number = number.strip()

        matches = reg.findall(number)

        for match in matches:
            parts = [part for part in match if part]
            extracted_contents.append([name] + parts)
    return extracted_contents
````
